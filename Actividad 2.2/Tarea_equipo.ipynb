```python
#Importamos las librerias pandas y numpy
import pandas as pd
import numpy as np
```

```python
#Leer archivo de Ventas por Asesor
ventas = pd.read_excel("Ventas_por_Asesor.xlsx")

ventas.head()
```

```python
#Información general
ventas.info()
```

```python
#Identificar valores nulos por columna
ventas.isnull().sum()
```

```python
#Convertir espacios vacíos en valores nulos
ventas = ventas.replace(r'^\s*$', np.nan, regex=True)
```

```python
ventas.isnull().sum()
```

```python
#Crear copia del DataFrame
ventas2 = ventas.copy()
```

```python
#Sustituir valores nulos de ventas mensuales con 0
columnas_meses = ventas2.columns[1:31]

ventas2[columnas_meses] = ventas2[columnas_meses].fillna(0)
```

```python
#Sustituir valores nulos de la primera columna
ventas2[ventas2.columns[0]] = ventas2[
    ventas2.columns[0]
].fillna("NA")
```

```python
#Sustituir valores nulos de acumulados y rankings
columnas_finales = ventas2.columns[31:]

ventas2[columnas_finales] = ventas2[
    columnas_finales
].fillna("NA")
```

```python
#Comprobar valores nulos
ventas2.isnull().sum()
```

```python
#Total de valores nulos
ventas2.isnull().sum().sum()
```

```python
#Crear archivo CSV final
ventas2.to_csv(
    "Ventas_por_Asesor_Sin_Nulos.csv",
    index=False
)
```

```python
#Leer archivo Principales Canales
canales = pd.read_excel("Principales_Canales.xlsx")
canales.head()
```
